task 4:average function

```python
def average(numbers):
    return sum(numbers) / len(numbers)

marks = [78, 85, 90, 72, 95]

result = average(marks)

print("Average Marks:", result)
```

task 6:product sales dictionary

```python
# Exercise 6: Product Sales Dictionary

product_sales = {
    "Laptop": 10,
    "Mobile": 25,
    "Keyboard": 15,
    "Mouse": 20
}

for product, units in product_sales.items():
    print(product, ":", units, "units")

total_units = sum(product_sales.values())

print("Total Units Sold:", total_units)
```

task 7:monthly sales count

```python
# Exercise 7: Count months with sales above 50

monthly_sales = [45, 60, 75, 30, 80, 55, 40, 90, 65, 20, 70, 50]

count = 0

for sales in monthly_sales:
    if sales > 50:
        count += 1

print("Months with sales above 50:", count)
```

task 8:grade

```python
# Exercise 8: Calculate Grade

marks = float(input("Enter your marks (0-100): "))

if marks < 0 or marks > 100:
    print("Invalid marks")
elif marks >= 80:
    print("Grade: A")
elif marks >= 70:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
else:
    print("Grade: F")
```

task 9:Pandas DataFrame

```python
# Exercise 9: Create a Pandas DataFrame

import pandas as pd

monthly_sales = [45, 60, 75, 30, 80, 55, 40, 90, 65, 20, 70, 50]

months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

df = pd.DataFrame({
    "Month": months,
    "Sales": monthly_sales
})

print(df)

print("\nTotal Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Maximum Sales:", df["Sales"].max())
```

task 10:sales bar chart

```python
# Exercise 10: Draw a Bar Chart

import matplotlib.pyplot as plt

df.plot(x="Month", y="Sales", kind="bar")

plt.title("Monthly Sales Chart")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```
